```python
import os
import re
```

```python
for file in [f for f in os.listdir('./model_selection/jobs') if f.find('fastai') > 0]:
    match = re.search(r'project(j\d+_\d+)', file.split('.sh')[0])
    if match:
        project_name = match.group(1)
        if not f'lhs_model_selection_fastai_project_{project_name}.csv' in [f for f in os.listdir('./results') if f.find('fastai') > 0]:
            print(f'Running model_selection.py for fastai {project_name}')
            os.system(f'python ./model_selection/model_selection.py fastai {project_name}')
            
```
